```python
query="Buy mobile phone buy phone online"
#Convert to lowercase
query=query.lower()
#Remove punctuation (basic cleaning)
punctuation=".,!?;:"
for p in punctuation:
    query=query.replace(p, "")
#Split sentence into words
words=query.split()
#Count frequency using dictionary
count={}
for word in words:
    if word in count:
        count[word]+=1
    else:
        count[word]=1
#Display only words with count > 1
repeat_keywords={}
for word, count1 in count.items():
    if count1>1:
        repeat_keywords[word]=count1
print(repeat_keywords)
```
