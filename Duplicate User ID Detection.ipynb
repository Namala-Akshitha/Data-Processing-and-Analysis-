```python
user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
count = {}
#Counts occurrences of user_ids
for user in user_ids:
    if user in count:
        count[user] += 1
    else:
        count[user] = 1
#displays only user1 and user3
for user in ["user1", "user3"]:
    print(user, "→", count[user], "times")
```
