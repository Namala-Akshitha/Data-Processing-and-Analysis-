```python
emails = [
"ravi@gmail.com",
"anita@yahoo.com",
"kiran@gmail.com",
"suresh@gmail.com",
"meena@yahoo.com"
]
count={}
#splits and counts both the domains
for email in emails:
    domain = email.split("@")[1]
    if domain in count:
        count[domain] += 1
    else:
        count[domain] = 1
#calcualte the percentage usage of each domain.
total_emails=len(emails)
for domain,count in count.items():
    percentage=(count/total_emails)*100
    print(f"{domain}: {percentage:.0f}%")
```
