```python
sales = [1200, 1500, 900, 2200, 1400, 3000]
#calculate avg daily sales
avg_sales=sum(sales)/len(sales)
#spike threshold 30% more than average.
threshold=avg_sales+0.30
#compare each day to detect spike days
for s in range(len(sales)):
    if sales[s]>threshold:
       print(f"Day {s+1}: {sales[s]}")
```
